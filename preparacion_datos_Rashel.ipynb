## 6. Selección de variables

**Criterios de eliminación**
1. Vacía: 99% o más de las filas valen 0 y las filas con valor no se concentran en una clase.
2. Copia: igual a otra columna (o a un cálculo exacto de otras) en 95% o más de las filas, o con correlación de 0.95 o más. Si las filas distintas se concentran en una clase, solo se elimina si otra columna que se conserva ya marca esas filas.
3. Sin relación con los ataques: el valor es aleatorio.
4. Acumulada: suma o cuenta a lo largo del flujo (duración, totales y conteos de paquetes), así que su valor final solo se conoce cuando el ataque ya terminó. Se conserva la razón que la contiene (promedio por paquete, por segundo o vuelta/ida).

### Variables acumuladas

Se eliminan `duracion` y las variables que suman o cuentan a lo largo del flujo (totales y conteos de
paquetes), porque su valor final solo se conoce cuando el ataque ya terminó. En su lugar se conservan las
razones que las contienen (promedio por paquete, paquetes por segundo, razón vuelta/ida): son un total
dividido entre paquetes o entre segundos, resumen la misma información y no dependen de cuánto duró la
conexión, por lo que podrían estimarse con los primeros paquetes.

```python
def igual_a_razon(valor, numerador, denominador):
    validas = denominador > 1e-3
    return np.isclose(data.loc[validas, valor], (numerador / denominador)[validas], rtol=1e-3).mean() * 100

paquetes = data["ida_paquetes"] + data["vuelta_paquetes"]
razones = [
    ("razon_vuelta_ida", "vuelta_paquetes / ida_paquetes", data["vuelta_paquetes"], data["ida_paquetes"]),
    ("ida_paquetes_por_seg", "ida_paquetes / duracion", data["ida_paquetes"], data["duracion"]),
    ("flujo_bytes_por_seg", "flujo_carga_total / duracion", data["flujo_carga_total"], data["duracion"]),
    ("ida_carga_prom", "ida_carga_total / ida_paquetes", data["ida_carga_total"], data["ida_paquetes"]),
    ("vuelta_carga_prom", "vuelta_carga_total / vuelta_paquetes", data["vuelta_carga_total"], data["vuelta_paquetes"]),
    ("ida_intervalo_prom", "ida_intervalo_total / (ida_paquetes - 1)", data["ida_intervalo_total"], data["ida_paquetes"] - 1),
    ("vuelta_intervalo_prom", "vuelta_intervalo_total / (vuelta_paquetes - 1)", data["vuelta_intervalo_total"], data["vuelta_paquetes"] - 1),
    ("flujo_intervalo_prom", "flujo_intervalo_total / (paquetes - 1)", data["flujo_intervalo_total"], paquetes - 1),
]
for valor, formula, numerador, denominador in razones:
    print(f"{valor} = {formula}: {igual_a_razon(valor, numerador, denominador):.1f}%")
print()

for col in ["ida_intervalo_total", "vuelta_intervalo_total", "flujo_intervalo_total", "activo_total", "inactivo_total"]:
    print(f"correlación {col} con duracion: {data[col].corr(data['duracion']):.3f}")
```

##### **5. Evaluación final (común al árbol y al bosque)**

El modelo final se reentrena con entrenamiento + validación y se evalúa una sola vez en prueba. Se reportan las mismas métricas que en KNN y la red neuronal (exactitud, macro-F1, recall macro y recall de SSH y FIN), se guardan los resultados en CSV y se generan las mismas figuras para el paper.

```python
from pathlib import Path
from matplotlib import gridspec
from sklearn.metrics import precision_score, confusion_matrix

### Unir entrenamiento y validación para el modelo final
X_TV_cod = pd.concat([X_T_cod, X_V_cod], ignore_index=True)
Y_TV = pd.concat([Y_T, Y_V], ignore_index=True)

### Clases en el mismo orden que los códigos de KNN y la red neuronal
clases = sorted(Y_T.unique())
SSH, FIN = "Metasploit_Brute_Force_SSH", "NMAP_FIN_SCAN"

### Estilo de las figuras del paper (igual que KNN y la red neuronal)
ESTILO = {
    'font.family': 'DejaVu Sans',
    'font.size': 11,
    'axes.titlesize': 12,
    'axes.labelsize': 11,
    'axes.spines.top': False,
    'axes.spines.right': False,
    'figure.dpi': 150,
    'savefig.dpi': 300,
    'savefig.facecolor': 'white',
}
BLACK, GRAY, LIGHT = '#000000', '#444444', '#d9d9d9'


def evaluar(modelo, X, Y):
    pred = modelo.predict(X)
    recall = pd.Series(recall_score(Y, pred, labels=clases, average=None, zero_division=0), index=clases)
    return {
        "exactitud": accuracy_score(Y, pred),
        "f1_macro": f1_score(Y, pred, average="macro", zero_division=0),
        "recall_macro": recall_score(Y, pred, average="macro", zero_division=0),
        "recall_SSH": recall[SSH],
        "recall_FIN": recall[FIN],
    }


def resultados_finales(nombre, sufijo, modelo, PHI, busqueda):
    Y_pred_t = modelo.predict(X_t_cod)

    ### Métricas en T + V y en prueba
    metricas = pd.DataFrame({
        "entrenamiento + validación": evaluar(modelo, X_TV_cod, Y_TV),
        "prueba": evaluar(modelo, X_t_cod, Y_t),
    }).T
    display(metricas)
    print(classification_report(Y_t, Y_pred_t, labels=clases, digits=3, zero_division=0))

    ### Reporte por clase, matriz de confusión e importancia de variables
    reporte = pd.DataFrame(classification_report(Y_t, Y_pred_t, labels=clases,
                                                 output_dict=True, zero_division=0)).T.loc[clases]
    reporte["support"] = reporte["support"].astype(int)

    cm = confusion_matrix(Y_t, Y_pred_t, labels=clases)
    matriz = pd.DataFrame(cm, index=clases, columns=clases)
    display(matriz)

    importancias = pd.Series(modelo.feature_importances_, index=X_T_cod.columns).sort_values(ascending=False)
    display(importancias.head(10))

    ### Guardar resultados (mismo formato que resultados/knn/)
    carpeta = Path(f"resultados/{sufijo}")
    carpeta.mkdir(parents=True, exist_ok=True)
    hiper = {k: v for k, v in PHI.items() if k not in ["F1_T", "F1_V"]}
    metricas.assign(modelo=nombre, **hiper).rename_axis("conjunto").to_csv(carpeta / f"metricas_{sufijo}.csv")
    reporte.to_csv(carpeta / "reporte_por_clase_prueba.csv")
    matriz.to_csv(carpeta / "matriz_confusion_prueba.csv")
    busqueda.to_csv(carpeta / "busqueda_hiperparametros.csv")
    importancias.to_csv(carpeta / "importancia_variables.csv")

    ### Datos de SSH y FIN para las figuras
    i_ssh, i_fin = clases.index(SSH), clases.index(FIN)
    cm_ssh_fin = cm[np.ix_([i_ssh, i_fin], [i_ssh, i_fin])]
    n_ssh, n_fin = int((Y_t == SSH).sum()), int((Y_t == FIN).sum())

    acc = accuracy_score(Y_t, Y_pred_t)
    f1_macro = f1_score(Y_t, Y_pred_t, average="macro", zero_division=0)
    f1_ssh, f1_fin = f1_score(Y_t, Y_pred_t, labels=[SSH, FIN], average=None, zero_division=0)
    rec_ssh, rec_fin = recall_score(Y_t, Y_pred_t, labels=[SSH, FIN], average=None, zero_division=0)
    pre_ssh, pre_fin = precision_score(Y_t, Y_pred_t, labels=[SSH, FIN], average=None, zero_division=0)

    dir_figs = Path(f"resultados/{sufijo}")
    dir_figs.mkdir(parents=True, exist_ok=True)

    with plt.rc_context(ESTILO):
        ### Figura 1. Métricas globales
        fig, ax = plt.subplots(figsize=(7.0, 4.0))
        nombres_m = ['Exactitud global', 'F1 macro', 'F1 SSH', 'F1 FIN']
        vals = [acc, f1_macro, f1_ssh, f1_fin]
        x = np.arange(len(nombres_m))
        ax.bar(x, vals, width=0.55, color=[LIGHT, LIGHT, GRAY, GRAY], edgecolor=BLACK, linewidth=1.0)
        ax.set_ylim(0, 1.25)
        ax.set_ylabel('Valor')
        ax.set_xticks(x)
        ax.set_xticklabels(nombres_m)
        ax.set_title(f'{nombre} — métricas globales y F1 de SSH y FIN (prueba)')
        ax.axhline(1.0, color='#c8c8c8', linewidth=0.8, linestyle='--', zorder=0)
        for i, v in enumerate(vals):
            ax.text(i, min(v + 0.04, 1.18), f'{v:.3f}', ha='center', va='bottom', fontsize=11)

        ruta1 = dir_figs / 'fig1_metricas_globales.png'
        fig.savefig(ruta1, bbox_inches='tight', pad_inches=0.3)
        plt.show()
        print('Guardada:', ruta1)

        ### Figura 2. Matriz y métricas de SSH y FIN
        fig = plt.figure(figsize=(6.8, 6.2))
        gs = gridspec.GridSpec(2, 1, height_ratios=[1.2, 1.0], hspace=0.65)

        ax = fig.add_subplot(gs[0])
        ax.imshow(cm_ssh_fin, cmap='Greys', vmin=0, vmax=max(5, int(cm_ssh_fin.max())), aspect='equal')
        ax.set_xticks([0, 1], ['SSH', 'FIN'])
        ax.set_yticks([0, 1], ['SSH', 'FIN'])
        ax.set_xlabel('Predicho')
        ax.set_ylabel('Real')
        ax.set_title(f'{nombre} — matriz de confusión de SSH y FIN (prueba)')
        for i in range(2):
            for j in range(2):
                v = int(cm_ssh_fin[i, j])
                ax.text(j, i, str(v), ha='center', va='center', fontsize=16,
                        color='white' if v >= 3 else BLACK, fontweight='bold')
        for spine in ax.spines.values():
            spine.set_visible(True)
            spine.set_color(BLACK)
            spine.set_linewidth(1.0)

        ax2 = fig.add_subplot(gs[1])
        ax2.axis('off')
        ax2.set_title(f'{nombre} — métricas de SSH y FIN (prueba)', pad=10)
        cols = ['Clase', 'Soporte', 'Recall', 'Precision', 'F1']
        rows = [
            ['SSH', str(n_ssh), f'{rec_ssh:.3f}', f'{pre_ssh:.3f}', f'{f1_ssh:.3f}'],
            ['FIN', str(n_fin), f'{rec_fin:.3f}', f'{pre_fin:.3f}', f'{f1_fin:.3f}'],
        ]
        table = ax2.table(cellText=rows, colLabels=cols, loc='upper center', cellLoc='center',
                          colWidths=[0.18, 0.14, 0.16, 0.18, 0.14])
        table.auto_set_font_size(False)
        table.set_fontsize(11)
        table.scale(1.15, 2.2)
        for (r_, c_), cell in table.get_celld().items():
            cell.set_edgecolor('#555555')
            cell.set_linewidth(0.7)
            if r_ == 0:
                cell.set_facecolor('#e6e6e6')
                cell.get_text().set_fontweight('bold')
            else:
                cell.set_facecolor('white')

        ruta2 = dir_figs / 'fig2_matriz_y_metricas_ssh_fin.png'
        fig.savefig(ruta2, bbox_inches='tight', pad_inches=0.35)
        plt.show()
        print('Guardada:', ruta2)
```

##### **2. Optimización de hiperparámetros**

```python
# Definición de hiperparametros
PROFUNDIDAD = [5, 10, 20, None]       
OBS_HOJAS   = [1, 5, 20]              
OBS_DIVIDIR = [2, 10]                 
PESO_CLASES = [None, "balanced"]     
CRITERIOS   = ["gini", "entropy"]     

# Construir las configuraciones de hiperparametros
PHI = {}
j   = 0

X_TICKS = []

for p in PROFUNDIDAD:
    for h in OBS_HOJAS:
        for d in OBS_DIVIDIR:
            for w in PESO_CLASES:
                for c in CRITERIOS:
                    PHI[j] = {"profundidad":p, "obs_dividir":d, "obs_hoja":h,
                              "peso_clases":w, "criterio":c}
                    j     += 1

                    X_TICKS += [f"P:{p} D:{d} H:{h} W:{w} C:{c}"]

# Optimizar
F1_T = []
F1_V = []

RESULTS = PHI.copy()

for (j, PHI_j) in PHI.items():
    # Crear un modelo de dt
    f_dt = DecisionTreeClassifier(criterion=PHI_j['criterio'],
                                  max_depth=PHI_j['profundidad'],
                                  min_samples_split=PHI_j['obs_dividir'],
                                  min_samples_leaf=PHI_j['obs_hoja'],
                                  class_weight=PHI_j['peso_clases'],
                                  random_state=42)

    # Entrenar el modelo de dt
    f_dt.fit(X_T_cod, Y_T)

    # Realizar predicciones
    Y_pred_T = f_dt.predict(X_T_cod)
    Y_pred_V = f_dt.predict(X_V_cod)

    # Calcular el desempeño
    f1_T = f1_score(y_true=Y_T, y_pred=Y_pred_T, average="macro", zero_division=0)
    f1_V = f1_score(y_true=Y_V, y_pred=Y_pred_V, average="macro", zero_division=0)

    F1_T += [f1_T]
    F1_V += [f1_V]

    RESULTS[j]['F1_T'] = f1_T
    RESULTS[j]['F1_V'] = f1_V
# FIN_FOR_OPTIMIZAR
```

```python
plt.figure(figsize=(20,5))
plt.plot(range(j+1), F1_T, marker='o', color="blue", label="T")
plt.plot(range(j+1), F1_V, marker='o', color="orange", label="V")
plt.xticks(ticks=range(j+1), labels=X_TICKS, rotation=90)
plt.xlabel("PHI")
plt.ylabel("macro-F1")
plt.grid()
plt.legend()
plt.show()

r = pd.DataFrame(RESULTS).T
r["F1_T"] = r["F1_T"].astype(float)
r["F1_V"] = r["F1_V"].astype(float)
r = r.sort_values(by='F1_V', ascending=False)
display(r.head(10))

# Mejor macro-F1 de validación por criterio
display(r.groupby("criterio")["F1_V"].max())

# Mejor configuración según macro-F1 en V
j_mejor   = r.index[0]
PHI_mejor = PHI[j_mejor]
PHI_mejor
```

Con la optimización de hiperparámetros, el mejor árbol de decisión (criterio entropía, profundidad máxima 10, min_samples_split = 10, min_samples_leaf = 1, sin ponderación de clases) alcanzó un macro-F1 de 0,961 en entrenamiento y de 0,962 en validación, frente a 0,998 y 0,951 del modelo de referencia. El mejor macro-F1 de validación con Gini fue de 0,961 (min_samples_leaf = 5), una diferencia de 0,0014 respecto de entropía. Entre las 10 mejores configuraciones solo una usó ponderación de clases.

##### **2. Optimización de hiperparámetros**

```python
# Definición de hiperparametros
PROFUNDIDAD = [10, None]                          
OBS_HOJAS   = [1, 5]                              
OBS_DIVIDIR = [2]                                 
N_ARBOLES   = [100]                               
N_ATRIBUTOS = [7]                                 
PESO_CLASES = [None, "balanced_subsample"]        
CRITERIOS   = ["gini", "entropy"]                 

# Construir las configuraciones de hiperparametros
PHI = {}
j   = 0

X_TICKS = []

for a in N_ARBOLES:
    for p in PROFUNDIDAD:
        for h in OBS_HOJAS:
            for d in OBS_DIVIDIR:
                for n in N_ATRIBUTOS:
                    for w in PESO_CLASES:
                        for c in CRITERIOS:
                            PHI[j] = {"n_arboles":a, "profundidad":p, "obs_dividir":d, "obs_hoja":h,
                                      "n_atributos":n, "peso_clases":w, "criterio":c}
                            j     += 1

                            X_TICKS += [f"A:{a} P:{p} D:{d} H:{h} F:{n} W:{w} C:{c}"]

# Optimizar
F1_T = []
F1_V = []

RESULTS = PHI.copy()

for (j, PHI_j) in PHI.items():
    # Crear un modelo de rf
    f_rf = RandomForestClassifier(criterion=PHI_j['criterio'],
                                  n_estimators=PHI_j['n_arboles'],
                                  max_depth=PHI_j['profundidad'],
                                  min_samples_split=PHI_j['obs_dividir'],
                                  min_samples_leaf=PHI_j['obs_hoja'],
                                  max_features=PHI_j['n_atributos'],
                                  class_weight=PHI_j['peso_clases'],
                                  n_jobs=-1,
                                  random_state=42)

    # Entrenar el modelo de rf
    f_rf.fit(X_T_cod, Y_T)

    # Realizar predicciones
    Y_pred_T = f_rf.predict(X_T_cod)
    Y_pred_V = f_rf.predict(X_V_cod)

    # Calcular el desempeño
    f1_T = f1_score(y_true=Y_T, y_pred=Y_pred_T, average="macro", zero_division=0)
    f1_V = f1_score(y_true=Y_V, y_pred=Y_pred_V, average="macro", zero_division=0)

    F1_T += [f1_T]
    F1_V += [f1_V]

    RESULTS[j]['F1_T'] = f1_T
    RESULTS[j]['F1_V'] = f1_V
# FIN_FOR_OPTIMIZAR

plt.figure(figsize=(15,5))
plt.plot(range(j+1), F1_T, marker='o', color="blue", label="T")
plt.plot(range(j+1), F1_V, marker='o', color="orange", label="V")
plt.xticks(ticks=range(j+1), labels=X_TICKS, rotation=90)
plt.xlabel("PHI")
plt.ylabel("macro-F1")
plt.grid()
plt.legend()
plt.show()

r = pd.DataFrame(RESULTS).T
r["F1_T"] = r["F1_T"].astype(float)
r["F1_V"] = r["F1_V"].astype(float)
r = r.sort_values(by='F1_V', ascending=False)
display(r.head(10))

# Mejor macro-F1 de validación por criterio
display(r.groupby("criterio")["F1_V"].max())

# Mejor configuración según macro-F1 en V
j_mejor   = r.index[0]
PHI_mejor = PHI[j_mejor]
PHI_mejor
```

Se evaluaron 16 configuraciones de bosque aleatorio mediante macro-F1 en validación. La configuración 10 obtuvo el mejor resultado (0,9731): 100 árboles, profundidad ilimitada, una observación mínima por hoja, 7 atributos por división, ponderación balanced_subsample y criterio Gini. Le siguieron la configuración 8 (0,9728), igual pero sin ponderación, y la 9 (0,9725), con entropía. Las tres primeras difieren en menos de 0,001.
