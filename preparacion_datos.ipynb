## 6. Selección de variables

**Criterios de eliminación**
1. Vacía: 99% o más de las filas valen 0 y las filas con valor no se concentran en una clase.
2. Copia: igual a otra columna (o a un cálculo exacto de otras) en 95% o más de las filas.
3. Sin relación con los ataques: el valor es aleatorio.

### Grupo 1 - Conexión

```python
print(f"puerto_origen valores distintos: {data['puerto_origen'].nunique():,} de {len(data):,} filas")
```

### Grupo 2 - Tiempo

```python
copia = (data["flujo_intervalo_total"] - data["duracion"] * 1_000_000).abs() < 1
print(f"flujo_intervalo_total = duracion x 1,000,000: {copia.mean() * 100:.1f}%")
```

```python
con_valor = data["inactivo_total"] != 0
for col in ["inactivo_min", "inactivo_max", "inactivo_prom"]:
    mismas_filas = ((data[col] != 0) == con_valor).mean() * 100
    igual = (data.loc[con_valor, col] == data.loc[con_valor, "inactivo_total"]).mean() * 100
    print(f"{col}: mismas filas con valor {mismas_filas:.1f}% | igual a inactivo_total {igual:.1f}%")
```

### Grupo 3 - Cantidad y tamaño

```python
copia_total = data["flujo_carga_total"] == data["ida_carga_total"] + data["vuelta_carga_total"]
copia_max = data["flujo_carga_max"] == data[["ida_carga_max", "vuelta_carga_max"]].max(axis=1)
copia_min = data["flujo_carga_min"] == data[["ida_carga_min", "vuelta_carga_min"]].min(axis=1)
print(f"flujo_carga_total = ida + vuelta: {copia_total.mean() * 100:.1f}%")
print(f"flujo_carga_max = mayor entre ida y vuelta: {copia_max.mean() * 100:.1f}%")
print(f"flujo_carga_min = menor entre ida y vuelta: {copia_min.mean() * 100:.1f}%")
```

### Grupo 4 - Velocidad

```python
suma = np.isclose(data["flujo_paquetes_por_seg"], data["ida_paquetes_por_seg"] + data["vuelta_paquetes_por_seg"])
producto = np.isclose(data["vuelta_paquetes_por_seg"], data["ida_paquetes_por_seg"] * data["razon_vuelta_ida"])
print(f"flujo_paquetes_por_seg = ida + vuelta: {suma.mean() * 100:.1f}%")
print(f"vuelta_paquetes_por_seg = ida x razon_vuelta_ida: {producto.mean() * 100:.1f}%")
```

### Grupo 5 - Ráfagas

```python
for col in ["ida_bloque_bytes", "ida_bloque_paquetes", "ida_bloque_tasa",
            "vuelta_bloque_bytes", "vuelta_bloque_paquetes", "vuelta_bloque_tasa"]:
    por_clase = (data[col] != 0).groupby(data["tipo_ataque"]).mean() * 100
    print(f"{col}: {(data[col] == 0).mean() * 100:.1f}% valen 0 | máximo con valor en una clase: {por_clase.max():.1f}%")
```

```python
distintas_ref = data["ida_subflujo_paquetes"] != data["ida_paquetes"]
pares = [
    ("vuelta_subflujo_paquetes", "vuelta_paquetes"),
    ("ida_subflujo_bytes", "ida_carga_total"),
    ("vuelta_subflujo_bytes", "vuelta_carga_total"),
]
for subflujo, total in pares:
    distintas = data[subflujo] != data[total]
    print(f"{subflujo} = {total}: {(~distintas).mean() * 100:.1f}%")
    print(f"   filas distintas que también lo son en ida_subflujo_paquetes: {distintas_ref[distintas].mean() * 100:.1f}%")
```

### Grupo 6 - Control TCP

```python
print(f"vuelta_bandera_URG: {(data['vuelta_bandera_URG'] == 0).mean() * 100:.1f}% valen 0")

cwr = data["bandera_CWR"] != 0
ece = data["bandera_ECE"] != 0
print(f"bandera_CWR: {(~cwr).mean() * 100:.1f}% valen 0 | {cwr.sum()} filas con valor:",
      data.loc[cwr, "tipo_ataque"].value_counts().to_dict())
print(f"bandera_ECE: {(~ece).mean() * 100:.1f}% valen 0 | {ece.sum()} filas con valor, {(cwr & ece).sum()} en las mismas filas que bandera_CWR")
```
